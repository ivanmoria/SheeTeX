```python
import pandas as pd
import numpy as np
import mca, os
```

```python
desktop_path = os.path.expanduser("~/Desktop")
counts = pd.read_csv(
    '/Users/ivanmoria/Desktop/17-16.csv',
    usecols=['Public', 'Design']
)

# Contar combinações repetidas e ordenar em ordem decrescente
grouped_counts = counts.groupby(['Public', 'Design']).size().reset_index(name='Count')
grouped_counts = grouped_counts.sort_values(by='Count', ascending=False)

# Mostrar resultado
print(grouped_counts)
```

```python
counts['Public'] = counts['Public'].astype(str).str.strip().str.title()
counts['Design'] = counts['Design'].astype(str).str.strip().str.title()

# Contar valores únicos de cada coluna separadamente
public_counts = counts['Public'].value_counts().reset_index()
public_counts.columns = ['Public', 'Count']
public_counts = public_counts.sort_values(by='Count', ascending=False)

design_counts = counts['Design'].value_counts().reset_index()
design_counts.columns = ['Design', 'Count']
design_counts = design_counts.sort_values(by='Count', ascending=False)

# Exibir resultados
print("=== Contagem de 'Public' ===")
print(public_counts)

print("\n=== Contagem de 'Design' ===")
print(design_counts)
```

```python
# === SALVAR COMO CSV NA ÁREA DE TRABALHO ===
public_file = os.path.join(desktop_path, 'public_counts.csv')
design_file = os.path.join(desktop_path, 'design_counts.csv')

public_counts.to_csv(public_file, index=False)
design_counts.to_csv(design_file, index=False)

print(f"\nArquivos salvos na área de trabalho:\n- {public_file}\n- {design_file}")
```
